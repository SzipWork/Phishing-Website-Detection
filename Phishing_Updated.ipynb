```python
#Dropping the Domain column
data = data0.drop(['Domain'], axis = 1).copy()
```

```python
#checking the data for null or missing values
data.isnull().sum()
```

```python
# shuffling the rows in the dataset so that when splitting the train and test set are equally distributed
data = data.sample(frac=1).reset_index(drop=True)
data.head()
```

```python
# Sepratating & assigning features and target columns to X & y
y = data['Label']
X = data.drop('Label',axis=1)
X.shape, y.shape
```

```python
# Splitting the dataset into train and test sets: 80-20 split
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 12)
X_train.shape, X_test.shape
```

```python
#importing packages
from sklearn.metrics import accuracy_score
```

# Decision Tree

```python

# Decision Tree model
from sklearn.tree import DecisionTreeClassifier

# instantiate the model
tree = DecisionTreeClassifier(max_depth = 5)
# fit the model
tree.fit(X_train, y_train)
```

```python
#predicting the target value from the model for the samples
y_test_tree = tree.predict(X_test)
y_train_tree = tree.predict(X_train)
```

```python
#computing the accuracy of the model performance
acc_train_tree = accuracy_score(y_train,y_train_tree)
acc_test_tree = accuracy_score(y_test,y_test_tree)

print("Decision Tree: Accuracy on training Data: {:.3f}".format(acc_train_tree))
print("Decision Tree: Accuracy on test Data: {:.3f}".format(acc_test_tree))
```

```python
#checking the feature improtance in the model
plt.figure(figsize=(9,7))
n_features = X_train.shape[1]
plt.barh(range(n_features), tree.feature_importances_, align='center')
plt.yticks(np.arange(n_features), X_train.columns)
plt.xlabel("Feature importance")
plt.ylabel("Feature")
plt.show()
```

```python
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense

# Define the Autoencoder Neural Network model (ANC)
input_dim = X_train.shape[1]
encoding_dim = 16  # You can adjust this dimension

input_layer = Input(shape=(input_dim, ))
encoder = Dense(encoding_dim, activation="relu")(input_layer)
decoder = Dense(input_dim, activation="sigmoid")(encoder)

anc = Model(inputs=input_layer, outputs=decoder)

anc.compile(optimizer='adam', loss='mean_squared_error')

# Train the ANC model (usually trained on the training data itself to learn the data distribution)
anc.fit(X_train, X_train, epochs=50, batch_size=32, shuffle=True, validation_split=0.2)

# Use the trained encoder as the feature extractor for a classifier
encoder_model = Model(inputs=input_layer, outputs=encoder)

# Get the encoded representation of the data
X_train_encoded = encoder_model.predict(X_train)
X_test_encoded = encoder_model.predict(X_test)

# Now, train a simple classifier (like a logistic regression or a small MLP) on the encoded data
# For simplicity, let's use a small MLP here
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

classifier = Sequential()
classifier.add(Dense(32, activation='relu', input_shape=(encoding_dim,)))
classifier.add(Dense(1, activation='sigmoid')) # Assuming binary classification (0 or 1)

classifier.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# Train the classifier on the encoded data
classifier.fit(X_train_encoded, y_train, epochs=50, batch_size=32, validation_split=0.2)

# Evaluate the classifier on the test data
loss, acc_test_anc = classifier.evaluate(X_test_encoded, y_test, verbose=0)
acc_train_anc = classifier.evaluate(X_train_encoded, y_train, verbose=0)[1]

print("ANC Classifier (on encoded data): Accuracy on training Data: {:.3f}".format(acc_train_anc))
print("ANC Classifier (on encoded data): Accuracy on test Data: {:.3f}".format(acc_test_anc))

storeResults('ANC Classifier', acc_train_anc, acc_test_anc)
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score

evaluation_results = []

# Function to calculate and store evaluation metrics
def storeEvaluationResults(model_name, y_true, y_pred):
    precision = precision_score(y_true, y_pred)
    recall = recall_score(y_true, y_pred)
    f1 = f1_score(y_true, y_pred)
    evaluation_results.append({
        'Model': model_name,
        'Precision': precision,
        'Recall': recall,
        'F1 Score': f1
    })

# Calculate metrics for each model
storeEvaluationResults('Decision Tree', y_test, y_test_tree)
storeEvaluationResults('Random Forest', y_test, y_test_forest)
storeEvaluationResults('XGBoost', y_test, y_test_xgb)
storeEvaluationResults('SVM', y_test, y_test_svm)
storeEvaluationResults('KNN', y_test, y_test_knn)
storeEvaluationResults('Multilayer Perceptrons', y_test, y_test_mlp)

# For ANC Classifier, use the predictions obtained previously
storeEvaluationResults('ANC Classifier', y_test, y_test_anc_pred)


# Create a DataFrame for evaluation results
evaluation_df = pd.DataFrame(evaluation_results)

# Display the evaluation results
print("\nModel Evaluation Metrics:")
evaluation_df
```

```python
# Pipeline
!pip install googlesearch-python

import re
from urllib.parse import urlparse
import requests
from bs4 import BeautifulSoup
import ipaddress
from googlesearch import search

def getFeatures(url):
    features = []
    # Domain - not used as it was dropped from training data
    # features.append(urlparse(url).netloc)

    # Have_IP
    try:
        ipaddress.ip_address(url)
        features.append(1)
    except:
        features.append(0)

    # Have_At
    if '@' in url:
        features.append(1)
    else:
        features.append(0)

    # URL_Length
    features.append(len(url))

    # URL_Depth
    features.append(url.count('/'))

    # Redirection
    try:
        response = requests.get(url, allow_redirects=False)
        if response.status_code in [301, 302]:
            features.append(1)
        else:
            features.append(0)
    except:
        features.append(0)

    # https_Domain
    if 'https' in urlparse(url).netloc:
        features.append(1)
    else:
        features.append(0)

    # TinyURL
    shortening_services = r"bit\.ly|goo\.gl|shorte\.st|go2l\.ink|x\.co|ow\.ly|t\.co|tinyurl|tr\.im|is\.gd|cli\.gs|" \
                          r"yfrog\.com|scrnch\.me|ff\.im|nr-act\.co\.za|wp\.me|peaksd\.com|" \
                          r"je\.suis\.gp|su\.pr|j\.mp|migre\.me"
    if re.search(shortening_services, url):
        features.append(1)
    else:
        features.append(0)

    # Prefix/Suffix
    if re.findall(r"-", url):
        features.append(1)
    else:
        features.append(0)

    # DNS_Record - This is difficult to implement reliably without external services or root access
    # We'll assume it's available for now (dummy value)
    features.append(1)

    # Web_Traffic - This requires external services like Alexa or similar
    # We'll assume it's available for now (dummy value)
    features.append(1)

    # Domain_Age - Requires external WHOIS lookup
    # We'll assume it's available for now (dummy value)
    features.append(1)

    # Domain_End - Requires external WHOIS lookup
    # We'll assume it's available for now (dummy value)
    features.append(1)

    # iFrame
    try:
        response = requests.get(url)
        soup = BeautifulSoup(response.content, 'html.parser')
        if soup.find_all('iframe'):
            features.append(1)
        else:
            features.append(0)
    except:
        features.append(0)

    # Mouse_Over
    try:
        response = requests.get(url)
        if re.findall("<script>.+onmouseover.+</script>", response.text):
            features.append(1)
        else:
            features.append(0)
    except:
        features.append(0)

    # Right_Click
    try:
        response = requests.get(url)
        if re.findall(r"event.button ?== ?2", response.text):
            features.append(1)
        else:
            features.append(0)
    except:
        features.append(0)

    # Web_Forwards - This is hard to determine programmatically from the URL alone
    # We'll assume it's available for now (dummy value)
    features.append(0) # Assuming no web forwards for simplicity

    return features

def predict_phishing(url):
    # Extract features from the input URL
    url_features = getFeatures(url)
    url_features_df = pd.DataFrame([url_features], columns=X_train.columns)

    # Make prediction using the best model (XGBoost based on previous code)
    prediction = xgb.predict(url_features_df)

    if prediction[0] == 0:
        return "Legitimate"
    else:
        return "Phishing"

# Example Usage:
url_to_check = input("Enter the URL to check: ")
prediction = predict_phishing(url_to_check)
print(f"The URL '{url_to_check}' is predicted as: {prediction}")
```
